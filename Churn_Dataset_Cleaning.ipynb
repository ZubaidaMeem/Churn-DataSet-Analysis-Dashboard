Import Libraries

```python
import pandas as pd
import numpy as np
```

Load Dataset

```python
df = pd.read_excel("./Unclean_Churn_Data.xlsx")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

Check Missing Value

```python
df.isnull().sum()
```

Check Duplicate Records

```python
df.duplicated().sum()
```

Remove Duplicate Record

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

Replace Dirty Values

```python
df.replace(['N/A','NULL','', ' '], np.nan, inplace=True)
```

```python
df.isnull().sum()
```

Remove Extra Spaces

```python
df = df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

Proper Case

```python
df["Customer_Name"] = df["Customer_Name"].str.title()
df["Gender"] = df["Gender"].str.title()
# df["State"] = df["State"].str.title()
df["City"] = df["City"].str.title()
df["Subscription_Type"] = df["Subscription_Type"].str.title()
df["Contract_Type"] = df["Contract_Type"].str.title()
df["Churn"] = df["Churn"].str.title()

# df = df.apply(lambda x:x.str.title() if x.dtype=="object" else x)
#  customer_id would auto apply


df["State"] = df["State"].apply(lambda x: x.upper() if isinstance(x, str) and len(x) <= 3 else x.title() if isinstance(x, str) else x)
```

Standardize Churn Column

```python
df["Churn"].value_counts()
```

Convert Numeric Columns

```python
num_cols = ["Age", "Tenure_Months", "Monthly_Charges","Total_Charges"]

for i in num_cols:
    df[i] = pd.to_numeric(df[i],errors="coerce")
```

```python
df.describe()
```

Remove Invalid Age

```python
df = df[(df["Age"] >= 18) & (df["Age"] <= 100)]
```

Remove Negative Data

```python
df = df[(df["Monthly_Charges"] >= 0)]
df = df[(df["Total_Charges"] >= 0)]
```

```python
df.shape
```

Convert Date

```python
df["Last_Interaction_Date"] = pd.to_datetime(df["Last_Interaction_Date"],errors="coerce")
```

Fill Missing Values

```python
df["Tech_Support"] = df["Tech_Support"].fillna("No")
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
df["Monthly_Charges"] = df["Monthly_Charges"].fillna(df["Monthly_Charges"].mean())
df["Tenure_Months"] = df["Tenure_Months"].fillna(df["Tenure_Months"].mean())
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

Feature Engineering
Customer Lifetime Value

```python
df["Customer_Value"] = df["Monthly_Charges"] * df["Tenure_Months"]
```

```python
df.head()
```

Monthly Revenue

```python
df["Monthly_Revenue"] = df["Monthly_Charges"]
```

Tenure Group "0-12", "13-24","25-48","49-72"

```python
bins =[0,12,24,48,72]
label = labels = ["0-12 month", "13-24 month", "25-48 month", "49-72 month"]
df["Tenure_Group"] = pd.cut(df["Tenure_Months"], bins=bins,labels=labels)
```

```python
df.head()
```

Senior Citizen Flag

```python
df["Senior_Flag"] = np.where(df["Age"] >= 60, "Senior", "Adult")
```

```python
df.head()
```

Churn Flag

```python
df["Churn_Flag"] = df["Churn"].map({"Yes":1, "No" :0})
```

```python
df.head()
```

Final CHeck

```python
df.info()
```

```python
df["Payment_Method"].value_counts()
```

```python
df["Last_Interaction_Date"] = df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median())
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

Removing Missing Churn Flag datas

```python
df = df.dropna(subset=["Churn_Flag"])
```

```python
df.isnull().sum()
```

```python
df.shape
```

Clean Dataset

```python
df.to_csv(
    "Clean_Churn_Data.csv",
    index=False
)
```

Export into MySQL

```python
from dotenv import load_dotenv
import os
from sqlalchemy import create_engine

load_dotenv()
password = os.getenv("MYSQL_PASSWORD")
engine = create_engine(f"mysql+pymysql://root:{password}@localhost/churndb")
```

```python
df.to_sql(
name="customer_churn",
con=engine,
if_exists="replace",
index=False
)
```
